```python
import pandas as pd

data = {
    'Name': ['Arun', 'Bala', 'Cathy', 'Divya', 'Ezhil'],
    'Maths': [85, 90, 78, 95, 88],
    'Science': [80, 85, 90, 92, 86],
    'English': [75, 88, 82, 90, 80]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Calculate average marks

df['Average'] = df[['Maths', 'Science', 'English']].mean(axis=1)

print(df)
```

```python
# Find highest and lowest average marks

print("Highest Average:")
print(df.loc[df['Average'].idxmax()])

print("\nLowest Average:")
print(df.loc[df['Average'].idxmin()])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(df['Name'], df['Average'])

plt.xlabel('Student Name')
plt.ylabel('Average Marks')
plt.title('Student Average Marks Analysis')

plt.show()
```

```python
# Subject-wise average marks

subject_avg = df[['Maths', 'Science', 'English']].mean()

print("Subject-wise Average Marks:")
print(subject_avg)

print("\nBest Performing Subject:")
print(subject_avg.idxmax())
```

```python
plt.figure(figsize=(7,5))

plt.bar(subject_avg.index, subject_avg.values)

plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.title('Subject-wise Performance Analysis')

plt.show()
```
